```python
# Retail Sales Performance Analysis

## Objective

### Analyse retail sales data to identify product, regional and salesperson performance and provide useful business insights.

import pandas as pd
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Date": [
        "2025-01-05", "2025-01-12", "2025-01-20",
        "2025-02-03", "2025-02-14", "2025-02-25",
        "2025-03-04", "2025-03-15", "2025-03-28",
        "2025-04-06", "2025-04-18", "2025-04-27",
        "2025-05-08", "2025-05-19", "2025-05-29",
        "2025-06-07", "2025-06-17", "2025-06-28"
    ],

    "Region": [
        "North", "South", "East",
        "West", "North", "South",
        "East", "West", "North",
        "South", "East", "West",
        "North", "South", "East",
        "West", "North", "South"
    ],

    "Product": [
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Tablet",
        "Laptop", "Phone", "Tablet"
    ],

    "Salesperson": [
        "Alice", "Bob", "Charlie",
        "David", "Alice", "Bob",
        "Charlie", "David", "Alice",
        "Bob", "Charlie", "David",
        "Alice", "Bob", "Charlie",
        "David", "Alice", "Bob"
    ],

    "Units": [
        12, 20, 15,
        10, 25, 18,
        14, 22, 17,
        11, 28, 20,
        16, 24, 19,
        13, 30, 21
    ],

    "Revenue": [
        14400, 16000, 7500,
        12000, 20000, 9000,
        16800, 17600, 8500,
        13200, 22400, 10000,
        19200, 19200, 9500,
        15600, 24000, 10500
    ]
})

sales

```

```python
## Data Visualisation and Business Insights

Sales_by_product.plot(kind="bar")

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue £")
plt.show()
```

```python
Sales_by_region = (
    sales.groupby("Region")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
Sales_by_region.plot(kind="bar")

plt.title("Revenue by region")
plt.xlabel("Region")
plt.ylabel("Revenue £")
plt.show()
```

```python
Sales_by_salesperson = (sales.groupby("Salesperson")["Revenue"].sum()).sort_values(ascending=False)

Sales_by_salesperson.plot(kind="bar")

plt.title("Revenue by Salesperson")
plt.xlabel("Salesperson")
plt.ylabel("Revenue £")
plt.show()
```

```python
sales["Date"] = pd.to_datetime(sales["Date"])
```

```python
revenue_over_time = sales.groupby("Date")["Revenue"].sum()
```

```python
revenue_over_time .plot(kind="line")

plt.title("Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue £")
plt.show()
```

## Key Business Insights

1. **Product Performance:** Phones generated the highest total revenue at £119,200.

2. **Regional Performance:** The North region generated the highest total revenue at £86,100.

3. **Salesperson Performance:** Alice generated the highest total revenue at £86,100.

## Business Recommendations

1. **Prioritise Phones:** Phones generated the highest revenue and sold the most units. The company should ensure strong phone inventory levels and consider increasing marketing around this product.

2. **Investigate Regional Performance:** The North generated the highest revenue. Management should investigate what is driving its stronger performance and determine whether successful approaches can be replicated in other regions.

3. **Learn From Top Performers:** Alice generated the highest total revenue. The company could examine her sales approach and share successful practices with other salespeople.

## Conclusion

This analysis examined retail sales performance across products, regions and salespeople.

Phones were the strongest-performing product by both revenue and units sold, while the North was the highest-revenue region. Alice generated the highest revenue among salespeople.

The analysis highlights areas where the company could focus its inventory, marketing and sales strategies to improve overall performance.
